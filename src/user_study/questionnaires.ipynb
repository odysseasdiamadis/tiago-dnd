```python
import pandas as pd 
import matplotlib.pyplot as plt


QUESTIONNAIRE_DATA = "Questionario Tiago (Risposte) - Risposte del modulo 1.tsv"
```

```python
df = pd.read_csv(QUESTIONNAIRE_DATA, sep="\t")

print(df.head())
```

```python

# --- 📊 Statistics ---
avg_age = df['Età'].mean()
std_age = df['Età'].std()

print(f"Average age: {avg_age:.2f}")
print(f"Standard deviation: {std_age:.2f}")

# --- 📈 Histogram ---
plt.figure(figsize=(7,5))
plt.hist(df['Età'], bins=range(20, 51, 5), color='skyblue', edgecolor='black')
plt.title('Age Distribution Histogram')
plt.xlabel('Age')
plt.ylabel('Frequency')

# Mark mean and std on the plot
plt.axvline(avg_age, color='red', linestyle='dashed', linewidth=2, label=f'Mean = {avg_age:.2f}')
plt.axvline(avg_age + std_age, color='green', linestyle='dotted', linewidth=2, label=f'+1 Std = {avg_age + std_age:.2f}')
plt.axvline(avg_age - std_age, color='green', linestyle='dotted', linewidth=2, label=f'-1 Std = {avg_age - std_age:.2f}')

plt.legend()
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
# Count how many times each age appears
sex_counts = df['Sesso'].value_counts()

# Plot a pie chart
plt.figure(figsize=(6,6))
plt.pie(sex_counts, labels=sex_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Sex')
plt.show()
```

```python
# Count frequency of each study title
study_counts = df['Titolo di studio'].value_counts()

# Plot as a bar chart
plt.figure(figsize=(7,5))
study_counts.plot(kind='bar', color='cornflowerblue', edgecolor='black')

plt.title('Distribution of Study Titles')
plt.xlabel('Study Title')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.7)
plt.show()
```
